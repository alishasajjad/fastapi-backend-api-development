# Why Backend Exists

Business logic should not remain on the frontend because it is insecure.

Example:

A banking app.

Suppose the frontend contains:

```
Balance = 100000
```

If users modify this value,

they could change it to

```
Balance = 1000000
```

This creates security issues.

Therefore,

Important business logic stays inside the backend.

Backend handles:

- Payment calculation
- Authentication
- Authorization
- Balance calculation
- Transactions

# Communication Between Frontend and Backend

Frontend communicates with Backend using

> **HTTP Protocol**

Communication follows:

```
Request
     ↓
Backend
     ↓
Response
```

### Login Request Example

```text
User Clicks Login
      ↓
Frontend Sends Request
      ↓
Backend Verifies credentials
      ↓
Backend Sends Response
      ↓
Frontend Displays Result
```
