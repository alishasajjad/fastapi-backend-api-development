# Creating Your First FastAPI Endpoint

After creating the `FastAPI` application, the next step is to create an API endpoint.

FastAPI supports different HTTP methods, such as:

- **GET** → Retrieve data
- **POST** → Create new data
- **PUT** → Update existing data
- **DELETE** → Remove data
- **PATCH** → Partially update data
- **HEAD** → Retrieve response headers only
- **OPTIONS** → Show supported HTTP methods

To create a GET endpoint, we use the `@app.get()` decorator.

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/get-todos")
def get_todos():
    print("Get Todos Called")
    return "Get Todos Called"
```

### Explanation

- `@app.get("/get-todos")` creates a **GET API endpoint**.
- `/get-todos` is called the **route (URL path)**.
- `def get_todos():` defines the Python function that will execute when this endpoint is requested.
- `print()` displays a message in the server console.
- `return` sends the response back to the client.

The function is **not called manually**. FastAPI automatically calls it whenever a client sends a GET request to `/get-todos`.

# How FastAPI Handles Requests

Normally, Python functions only run when we call them.

Example:

```python
def greet():
    return "Hello"

greet()
```

In FastAPI, we **do not call the function ourselves**.

Instead, the framework waits for an HTTP request.

When a client (such as Postman, a web browser, or a frontend application) sends a request to:

```
GET /get-todos
```

FastAPI automatically:

1. Receives the request.
2. Checks the request path.
3. Finds the matching route.
4. Calls the corresponding function.
5. Returns the function's output as an HTTP response.

### Request Flow

```
Frontend / Postman
        │
        ▼
HTTP Request
        │
        ▼
FastAPI Application
        │
        ▼
Matching Route
        │
        ▼
Python Function
        │
        ▼
Return Response
        │
        ▼
Frontend / Client
```

This is why API functions are different from normal Python functions—they are executed only when an HTTP request reaches the corresponding endpoint.

# Understanding Routes, Server, IP Address, and Port

Backend code does not run inside the user's browser.

Instead, it runs on a **server**, which is a computer responsible for processing requests and sending responses.

When a client wants to access an API, three things are required:

## 1. IP Address

The IP address identifies the machine where the backend application is running.

Example:

```
192.168.1.10
```

## 2. Port Number

A single server can run multiple applications simultaneously.

Each application listens on a different **port number**.

Example:

```
8000
8080
5000
3000
```

The combination of **IP Address + Port Number** identifies a specific running application.

Example:

```
http://192.168.1.10:8000
```

## 3. Route (Path)

Inside one application, there can be many API endpoints.

Each endpoint is identified by its route.

Examples:

```
/get-todos
/get-single-todo
/create-todo
/update-todo
/delete-todo
```

A complete API URL looks like this:

```
http://192.168.1.10:8000/get-todos
```

Where:

- `http://` → HTTP protocol
- `192.168.1.10` → Server IP Address
- `8000` → Port Number
- `/get-todos` → API Route

## Multiple Routes Example

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/get-todos")
def get_todos():
    return "All Todos"

@app.get("/get-single-todo")
def get_single_todo():
    return "Single Todo"
```

Each route must be unique for the same HTTP method.

For example:

- `GET /get-todos` 
- `GET /get-single-todo` 

If you want to use the same route, you must use a different HTTP method.

Example:

```python
@app.get("/todos")
def get_todos():
    return "Fetch Todos"

@app.post("/todos")
def create_todo():
    return "Create Todo"
```

Here, the route is the same (`/todos`), but the HTTP methods (`GET` and `POST`) are different, so both endpoints are valid.

# Running a FastAPI Application

Normally, a Python file is executed using:

```bash
python main.py
```

When executed this way:

- The file runs only once.

# Why APIs Need to Keep Running

A backend API must remain active because requests can arrive at any time.

For example:

- A frontend application sends a request.
- A mobile application sends a request.
- Postman sends a request.

Since we cannot predict when a request will arrive, the API server must stay running continuously.

# Uvicorn

To keep a FastAPI application running continuously, we use **Uvicorn**.

Uvicorn is an **ASGI server** responsible for:

- Starting the FastAPI application.
- Keeping the application running.
- Listening for incoming HTTP requests.
- Forwarding requests to FastAPI.
- Returning responses back to the client.

# Why Did We Get "Not Found"?

Sometimes, after adding a new endpoint, the response may be:

```
404 Not Found
```

This happens because **Uvicorn is still running the old version of the application**.

When the server starts, it loads the current code into memory.

If you modify the code afterwards, Uvicorn does **not** automatically detect those changes unless it is restarted or auto-reload is enabled.

---

# Restarting the Server

To load the latest changes:

1. Stop the running server.

```
Ctrl + C
```

2. Start it again.

```bash
poetry run dev
```

Now Uvicorn loads the updated version of `main.py`, including the newly added API endpoint.

---

# Automatically Reloading the Server

Restarting the server after every change is inconvenient during development.

Uvicorn provides a **reload** option that automatically reloads the application whenever the source code changes.

Update the `start()` function as follows:

```python
def start():
    uvicorn.run(
        "todos.main:app",
        host="127.0.0.1",
        port=8080,
        reload=True
    )
```

---

# What Does `reload=True` Do?

```python
reload=True
```

This enables **auto-reload**.

Whenever you:

- Modify the source code.
- Save the file.

Uvicorn automatically:

- Stops the old server.
- Reloads the application.
- Starts the updated server.

This means you no longer need to manually restart the server after every code change.

---

# Running the Server

Start the application:

```bash
poetry run dev
```

Now every saved change is automatically reflected in the running application.

# API Documentation

Every API requires the following information:

- HTTP Method
- Host/IP Address
- Port Number
- Route
- Response Format

Normally, developers document these details manually.

FastAPI automatically generates this documentation.

---

# Automatic Swagger Documentation

When the server is running, open:

```
http://127.0.0.1:8080/docs
```

FastAPI automatically generates an interactive documentation page.

No additional coding is required.

---

# What Does the Documentation Show?

The documentation displays:

- All available API endpoints.
- HTTP methods (GET, POST, PUT, DELETE, etc.).
- Route paths.
- Request parameters.
- Request body (if required).
- Response format.
- Status codes.

This makes it much easier for frontend and backend developers to work together.

---

# Example

Suppose the application contains:

```python
@app.get("/get-todo")
def get_todo():
    return "Get Todo Called"

@app.post("/get-todo")
def create_todo():
    return "Post Todo Called"

@app.put("/update-todo")
def update_todo():
    return "Update Todo Called"

@app.get("/")
def hello_world():
    return "Hello World"
```

After refreshing the `/docs` page, FastAPI automatically lists all four endpoints.

No manual documentation is needed.

---

# Testing APIs Directly from Swagger UI

FastAPI documentation also allows you to test APIs.

Steps:

1. Open:

```
http://127.0.0.1:8080/docs
```

2. Select an endpoint.

3. Click **Try it out**.

4. Click **Execute**.

FastAPI sends the request and displays:

- Request URL
- Response Body
- Status Code
- Response Headers

This means you can test APIs directly from the browser without using Postman.

---

# Example

For the endpoint:

```python
@app.put("/update-todo")
def update_todo():
    return "Update Todo Called"
```

Click **Execute**.

Response:

```
Update Todo Called
```

The response appears immediately in the Swagger interface.

FastAPI provides several benefits:

- Very little code is required to create APIs.
- Automatic API documentation.
- Built-in Swagger UI.
- Interactive API testing.
- Automatic request validation.
- High performance.

# Summary

- New API endpoints require the server to reload before changes become available.
- `reload=True` automatically reloads the server whenever the code changes.
- `127.0.0.1` (localhost) allows access only from the same machine.
- Other devices require a public or network IP address.
- Domain names map to IP addresses through DNS.
- FastAPI automatically generates API documentation at:

```
http://127.0.0.1:8080/docs
```

- Swagger UI allows developers to view and test APIs without Postman.
- FastAPI significantly simplifies API development by providing automatic documentation and interactive testing tools.
