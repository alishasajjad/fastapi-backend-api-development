# Path Variables

A **Path Variable** (also called a **Dynamic Path**) allows us to receive values directly from the URL.

Example Route

```python
@app.get("/todos/{id}")
```

Here:

- `/todos` → Fixed Route
- `{id}` → Dynamic Path Variable

The value inside `{}` is received from the frontend.

---

# Receiving a Path Variable

The same variable name must also be declared as a function parameter.

```python
@app.get("/todos/{id}")
def get_todo(id):
    return id
```

The variable name inside the route and inside the function **must match**.

---

# Example

```python
@app.get("/todos/{id}")
def get_todo(id):
    print(id)
    return id
```

If the request is:

```
GET : http://127.0.0.1:8080/todos/13
```

The function receives:

```python
id = 13
```

Response

```
13
```

```
15
```

```
Frontend
      │
      ▼
GET /todos/112
      │
      ▼
FastAPI
      │
      ▼
id = 112
      │
      ▼
Return 112
```

# Multiple Path Variables

You can receive more than one value.

```python
@app.get("/students/{username}/{roll_number}")
def get_student(username, roll_number):
    print(username)
    print(roll_number)

    return {
        "username": username,
        "roll_number": roll_number
    }
```

Request

```
GET /students/Alisha/112
```

Response

```json
{
    "username":"Alisha",
    "roll_number":"112"
}
```

Although possible, this approach becomes difficult when many values need to be passed.

# Type Hints

FastAPI supports Python type hints.

Example

```python
@app.get("/students/{username}/{roll_number}")
def get_student(username: str, roll_number: int):
    return {
        "username": username,
        "roll_number": roll_number
    }
```

Benefits

- Better validation
- Automatic documentation
- Cleaner code

# Limitations of Path Variables

Imagine an admission form with 100 fields.

Using path variables would look like:

```
/students/Ali/101/ComputerScience/20/...etc
```

This is difficult to read and maintain.

Therefore, another technique is preferred.

---

# Query Parameters

Query Parameters send data after the **?** symbol in the URL.

Unlike Path Variables, nothing is written inside the route.

Example

```python
@app.get("/student")
def get_student(username: str, roll_number: str):
    return {
        "username": username,
        "roll_number": roll_number
    }
```

---

## Calling the API

```
GET /student?username=Alisha&roll_number=112
```

Response

```json
{
    "username":"Alisha",
    "roll_number":"112"
}
```

---

# Query Parameter URL Structure

```
/student
    ?
username=Ali
    &
roll_number=101
```

- `?` starts query parameters.
- `&` separates multiple parameters.

---

# Path Variables vs Query Parameters

| Path Variables | Query Parameters |
|---------------|------------------|
| Data is part of the URL path | Data comes after `?` |
| Uses `{}` in the route | No `{}` in the route |
| Better for resource IDs | Better for filters and optional values |
| Example: `/students/101` | Example: `/students?roll_number=101` |

# Solution - Student API (In-Memory List)

## Student Data

```python
students = [
    {
        "userName": "Alisha",
        "rollNo": 112
    },
    {
        "userName": "Zarmeen",
        "rollNo": 373
    }
]
```

---

## 1. Get All Students

Returns the complete list of students.

```python
@app.get("/getStudent")
def getStudent():
    return students
```

---

## 2. Get Single Student (Path Variable)

Retrieves a student using the roll number from the URL path.

```python
@app.get("/getStudent/{rollNo}")
def getStudentByPath(rollNo: int):
    for student in students:
        if student["rollNo"] == rollNo:
            return student

    return {"message": "Student not found"}
```

### Example

```
GET /getStudent/112
```

---

## 3. Get Single Student (Query Parameter)

Retrieves a student using a query parameter.

```python
@app.get("/searchStudent")
def searchStudent(rollNo: int):
    for student in students:
        if student["rollNo"] == rollNo:
            return student

    return {"message": "Student not found"}
```

### Example

```
GET /searchStudent?rollNo=373
```

---

## 4. Add Student

Adds a new student to the list.

```python
@app.get("/addStudent")
def addStudent(userName: str, rollNo: int):
    global students

    students.append({
        "userName": userName,
        "rollNo": rollNo
    })

    return students
```

### Example

```
GET /addStudent?userName=Ali&rollNo=101
```

---

## 5. Update Student

Updates the username of an existing student.

```python
@app.put("/updateStudent")
def updateStudent(rollNo: int, userName: str):

    for student in students:
        if student["rollNo"] == rollNo:
            student["userName"] = userName
            return student

    return {"message": "Student not found"}
```

### Example

```
PUT /updateStudent?rollNo=112&userName=Ali
```

---

## 6. Delete Student

Deletes a student using the roll number.

```python
@app.delete("/deleteStudent")
def deleteStudent(rollNo: int):

    for student in students:
        if student["rollNo"] == rollNo:
            students.remove(student)
            return students

    return {"message": "Student not found"}
```

### Example

```
DELETE /deleteStudent?rollNo=112
```
