# What is an API?

**API (Application Programming Interface)** is a communication medium that allows two different software applications to communicate with each other.

### Example
- Frontend Application ↔ Backend Server
- Backend Server ↔ Third-party APIs

Whenever one application sends a request to another application and receives a response, communication happens through an **API**.

# Why Do Backend Developers Need API Testing?

Normally, requests come from the frontend.

However, a backend developer may not have the frontend application ready.

Instead of building the frontend just for testing, developers use an API testing tool.

---

# API Methods (HTTP Methods)

REST APIs use different HTTP methods.

Main methods are:

- GET
- POST
- PUT
- DELETE

## GET Method

Purpose:

Retrieve data from the server.

Example:

- Get all users
- Get products
- Get profile information

## POST Method

Purpose:

Send new data to the server.

Examples:

- Register a user
- Submit a form
- Create a new product

## PUT Method

Purpose:

Update existing data.

Example:

- Update user profile
- Update product details

## DELETE Method

Purpose:

Remove existing data.

Example:

- Delete a user
- Delete a product

# Less Common HTTP Methods

Other available methods include:

- PATCH
- HEAD
- OPTIONS

These methods are used in specific situations but are less common than GET, POST, PUT, and DELETE.

# 🌐 How a Backend Request Works (Story)

Imagine you build a **FastAPI backend** for an online shopping application.

A user opens your website and clicks the **Login** button.

The frontend cannot directly execute Python code because the backend is running on another machine. Instead, it sends an **HTTP Request** over the internet.

The backend processes the request, performs the required business logic, communicates with the database if needed, and finally sends an **HTTP Response** back to the frontend.

```text
User
   │
   ▼
Frontend
   │
HTTP Request
   │
Internet
   │
Backend Server
   │
Database
   │
HTTP Response
   │
Frontend
   │
User
```

# 🖥️ Step 1 — What is a Server?

A **Server** is a powerful computer that runs backend applications and serves multiple users simultaneously.

Unlike a personal laptop, a server has:

- High CPU Performance
- Large RAM
- Massive Storage
- Fast Network Connection
- 24/7 Availability

The backend of your application is deployed on a server so that users from anywhere in the world can access it.

# 🌍 Step 2 — What is an IP Address?

Every device connected to the internet has a unique **IP Address (Internet Protocol Address)**.

Example:

```text
192.168.1.10
```

The IP Address acts like the home address of a server.

Without an IP address, the internet would not know where to send the request.

# 🌐 Step 3 — What is a Domain Name?

Remembering IP addresses is difficult.

Instead of typing numbers, we use **Domain Names**.

Examples:

- google.com
- github.com
- amazon.com

A Domain Name is simply a human-readable name that points to an IP Address.

```text
google.com
      │
      ▼
142.250.xxx.xxx
```

This conversion is handled automatically by the **DNS (Domain Name System)**.

# 📡 Step 4 — What is HTTP?

HTTP stands for **HyperText Transfer Protocol**.

It is the communication protocol that allows the frontend and backend to exchange data.

```text
Frontend
      │
HTTP Request
      │
Backend
      │
HTTP Response
      │
Frontend
```

When communication is encrypted, HTTP becomes **HTTPS**.

# 🚪 Step 5 — What is a Port?

A single server can run multiple applications at the same time.

To identify the correct application, every application runs on a **Port Number**.

Example:

```text
192.168.1.10:8000
```

Here:

- **192.168.1.10** → Server IP Address
- **8000** → Port Number

Think of it like this:

- IP Address = Building Address
- Port = Apartment Number

The port tells the server which application should receive the request.

# 🛣️ Step 6 — What is an API Route (Endpoint)?

Inside one backend application, there are many Python functions.

Each function is connected to a unique **Route (Endpoint)**.

Examples:

```text
/login
/register
/products
/orders
```

When a request reaches the backend, FastAPI checks the route and executes the corresponding Python function.

```text
Request
    │
    ▼
/login
    │
    ▼
login() Function
    │
    ▼
Response
```

# 🔗 Step 7 — Understanding a Complete URL

Example URL:

```text
https://example.com:8000/users
```

This URL consists of four parts:

### Protocol

```text
https://
```

Defines how communication happens.

### Domain

```text
example.com
```

Human-readable address of the server.

### Port

```text
8000
```

Identifies the backend application running on the server.

### Route

```text
/users
```

Identifies the backend function that should execute.

```text
https://example.com:8000/users
│
├── https → Protocol
├── example.com → Domain
├── 8000 → Port
└── /users → API Route
```

# ⚙️ Step 8 — Backend Executes Business Logic

Once the request reaches the correct route, FastAPI executes the corresponding Python function.

The backend may:

- Validate user input
- Authenticate users
- Query the database
- Perform calculations
- Call third-party APIs
- Use AI models

Finally, it prepares a response for the client.

# 📤 Step 9 — Backend Returns the Response

After completing all processing, the backend sends a response back to the frontend.

```text
Request
   │
   ▼
Backend
   │
Business Logic
   │
Database
   │
JSON Response
   │
Frontend
   │
User
```

# 🛒 Real-World Example

Suppose a user opens the following URL:

```text
https://shop.com:8000/products
```

The request follows these steps:

1. The browser reads the URL.
2. DNS converts **shop.com** into an IP Address.
3. The request reaches the server.
4. Port **8000** identifies the FastAPI application.
5. The **/products** route is matched.
6. FastAPI executes the `get_products()` function.
7. The function retrieves product data from the database.
8. The backend returns a JSON response.
9. The frontend displays the products to the user.

# 📌 Summary

- **Server** → Powerful computer that hosts backend applications.
- **IP Address** → Unique numerical address of a server.
- **Domain Name** → Human-readable name that maps to an IP address.
- **HTTP/HTTPS** → Communication protocol between frontend and backend.
- **Port** → Identifies a specific application running on a server.
- **Route (Endpoint)** → Identifies which backend function should execute.
- **URL** → Combination of Protocol + Domain/IP + Port + Route.
- **FastAPI** → Receives requests, executes business logic, and returns responses.

# 📚 API Documentation

**API Documentation** is a guide that explains how to use an API correctly. It provides all the information developers need to communicate with the backend without reading its source code.

API documentation acts as a bridge between **Backend Developers** and **Frontend Developers**, making integration faster and easier.

# 🎯 Why API Documentation is Important

Without documentation, developers would have to ask the backend developer about every API endpoint.

Good API documentation helps developers:

- Understand available APIs
- Know which HTTP method to use
- Find the correct API URL
- Understand required request parameters
- View expected responses
- Test APIs easily
- Reduce development time

# 📄 What Does API Documentation Contain?

A complete API documentation usually includes:

- API Endpoint (URL)
- HTTP Method (GET, POST, PUT, DELETE)
- Request Parameters
- Request Body
- Response Format
- Status Codes
- Authentication Requirements
- Example Requests
- Example Responses
